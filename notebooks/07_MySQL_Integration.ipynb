# Notebook 7: MySQL Integration

Load the cleaned and forecast datasets into MySQL and execute business queries.

```python
import pandas as pd
from sqlalchemy import create_engine

# Update your credentials
user='root'
password='nazeera'
host='localhost'
database='retail_sales_db'

engine=create_engine(f"mysql+pymysql://{user}:{password}@{host}/{database}")

sales=pd.read_csv('feature_engineered_retail_sales.csv')
forecast=pd.read_csv('final_sales_forecast.csv')

sales.head()
```

```python
sales.to_sql('fact_sales',engine,if_exists='replace',index=False)
forecast.to_sql('sales_forecast',engine,if_exists='replace',index=False)
print('Tables created successfully.')
```

```python
queries={
'Total Sales':'SELECT SUM(Sales) AS Total_Sales FROM fact_sales;',
'Region Wise Sales':'SELECT Region,SUM(Sales) AS Sales FROM fact_sales GROUP BY Region;',
'Top 10 Products':'SELECT Product_Name,SUM(Sales) AS Sales FROM fact_sales GROUP BY Product_Name ORDER BY Sales DESC LIMIT 10;',
'Category Profit':'SELECT Category,SUM(Profit) AS Profit FROM fact_sales GROUP BY Category;',
'Forecast':'SELECT * FROM sales_forecast LIMIT 10;'
}
for title,q in queries.items():
    print('\n',title)
    print(pd.read_sql(q,engine))
```
