# Notebook 8: Power BI Dashboard Preparation

Prepare datasets, star schema, KPIs and DAX measures.

```python
import pandas as pd

df = pd.read_csv('feature_engineered_retail_sales.csv', parse_dates=['Order_Date', 'Ship_Date'])
df.head()
```

```python
# Create Dimension Tables
# Replace underscores with actual column names if needed
sub_cat_col = 'Sub_Category' if 'Sub_Category' in df.columns else 'Sub-Category'
prod_name_col = 'Product_Name' if 'Product_Name' in df.columns else 'Product Name'
cust_name_col = 'Customer_Name' if 'Customer_Name' in df.columns else 'Customer Name'
cust_id_col = 'Customer_ID' if 'Customer_ID' in df.columns else 'Customer ID'
prod_id_col = 'Product_ID' if 'Product_ID' in df.columns else 'Product ID'

dim_customer = df[[cust_id_col, cust_name_col, 'Segment']].drop_duplicates()
dim_product = df[[prod_id_col, prod_name_col, 'Category', sub_cat_col]].drop_duplicates()
dim_location = df[['State', 'City', 'Region']].drop_duplicates()

date_col = 'Order_Date' if 'Order_Date' in df.columns else 'Order Date'
date = pd.DataFrame({'Date': pd.date_range(df[date_col].min(), df[date_col].max())})
date['Year'] = date['Date'].dt.year
date['Month'] = date['Date'].dt.month
date['Month_Name'] = date['Date'].dt.month_name()
date['Quarter'] = date['Date'].dt.quarter

fact = df.copy()
print("Dimension and Fact tables created successfully!")
```

```python
dim_customer.to_csv('dim_customer.csv', index=False)
dim_product.to_csv('dim_product.csv', index=False)
dim_location.to_csv('dim_location.csv', index=False)
date.to_csv('dim_date.csv', index=False)
fact.to_csv('fact_sales.csv', index=False)
print('Star schema CSV files exported.')
```
